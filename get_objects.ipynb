```python
import numpy as np
import open3d as o3d
```

```python
data = np.load('/home/viny/run1/office/dense_pointcloud.npz')
```

```python
data.keys()
```

```python
for obj_id in np.unique(data['cluster_labels']):
    if obj_id == -1:
        continue    
    object_points = data['points'][data['cluster_labels'] == obj_id]
    object_colors = data['colors'][data['cluster_labels'] == obj_id]
    point_cloud = o3d.geometry.PointCloud()
    point_cloud.points = o3d.utility.Vector3dVector(object_points)
    point_cloud.colors = o3d.utility.Vector3dVector(object_colors)
    # Estimate normals
    point_cloud.estimate_normals(search_param=o3d.geometry.KDTreeSearchParamKNN(knn=30))
    normals = np.asarray(point_cloud.normals)
    object_data = np.hstack((object_points, normals))
    np.save(f'/home/viny/test_objects/{obj_id}.npy', object_data)
    
    o3d.io.write_point_cloud(f'/home/viny/test_objects/{obj_id}.ply', point_cloud)
```
